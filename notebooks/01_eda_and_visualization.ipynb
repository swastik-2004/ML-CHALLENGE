# Amazon ML Challenge 2026: Exploratory Data Analysis & Visualization

This notebook explores the Business Entity Resolution dataset across Source 1, Source 2, and Source 3.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
print("Libraries loaded successfully!")
```

## 1. Load Data Samples

```python
DATA_DIR = Path("../student_resource/dataset")

s1 = pd.read_csv(DATA_DIR / "train/train_source1.tsv", sep="\t", nrows=10000)
s2 = pd.read_csv(DATA_DIR / "train/train_source2.tsv", sep="\t", nrows=10000)
s3 = pd.read_csv(DATA_DIR / "train/train_source3.tsv", sep="\t", nrows=10000)
gt = pd.read_csv(DATA_DIR / "train/train_ground_truth.tsv", sep="\t", nrows=10000)

print(f"Source 1 sample: {s1.shape}")
print(f"Source 2 sample: {s2.shape}")
print(f"Source 3 sample: {s3.shape}")
print(f"Ground truth sample: {gt.shape}")
s1.head()
```

## 2. Match Count Analysis (Ground Truth)

```python
gt['matched_entity_ids'] = gt['matched_entity_ids'].fillna('')
gt['match_count'] = gt['matched_entity_ids'].apply(lambda x: len([m for m in x.split(',') if m.strip()]) if x else 0)

plt.figure(figsize=(10, 4))
sns.countplot(data=gt, x='match_count', palette='Blues_r')
plt.title("Distribution of Matches per Source 1 Entity in Sample")
plt.xlabel("Match Count")
plt.ylabel("Number of Entities")
plt.show()
```

## 3. Country Distribution

```python
test_s1 = pd.read_csv(DATA_DIR / "test/test_source1.tsv", sep="\t", nrows=10000)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
s1['country'].value_counts().plot(kind='bar', ax=axes[0], color=['#4A90E2', '#F5A623'])
axes[0].set_title("Train Countries (Source 1)")

test_s1['country'].value_counts().plot(kind='bar', ax=axes[1], color=['#4A90E2', '#F5A623', '#E94E77'])
axes[1].set_title("Test Countries (Source 1 - includes France)")
plt.tight_layout()
plt.show()
```
